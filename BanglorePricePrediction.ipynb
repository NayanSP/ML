```python
def ConvertRange(x):
    temp = x.split(' - ')
    if len(temp)==2:
        return (float(temp[0])+(float(temp[1])))/2
    try:
        return float(x)
    except:
        return None
```

```python
def remove_outlier_sqft(df):
    df_op = pd.DataFrame()
    for key, subdf in df.groupby('location'):
        m = np.mean(subdf.PricePerSqFt)
        sd= np.std(subdf.PricePerSqFt)
        gen_df = subdf[(subdf.PricePerSqFt>(m-sd)) & (subdf.PricePerSqFt<=(m+sd))]
        df_op = pd.concat([df_op, gen_df], ignore_index=True)
    return df_op

df = remove_outlier_sqft(df)
df.describe()
```

```python
def bhk_outlier_remover(df):
    exclude_index = np.array([])
    for loc, loc_df in df.groupby('location'):
        bhk_stats = {}
        for bhk, bhk_df in loc_df.groupby('bhk'):
            bhk_stats[bhk] = {
                'mean':np.mean(bhk_df.PricePerSqFt),
                'sd':np.std(bhk_df.PricePerSqFt),
                'cnt':bhk_df.shape[0]
            }
        print(loc, bhk_stats)
        for bhk, bhk_df in loc_df.groupby('bhk'):
            stats = bhk_stats.get(bhk-1)
            if stats and stats['cnt']>5:
                exclude_index = np.append(exclude_index, bhk_df[bhk_df.PricePerSqFt<stats['mean']].index.values)
    return df.drop(exclude_index, axis = 'index')
```

```python
df = bhk_outlier_remover(df)
df
```

```python
x = df.drop(columns='price')
y = df.price
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, Lasso, Ridge
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import make_column_transformer
from sklearn.pipeline import make_pipeline
from sklearn.metrics import r2_score
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.25, random_state=42)
```

```python
column_tran = make_column_transformer((OneHotEncoder(),
                                       ['location']),
                                      remainder='passthrough' )
```

```python
scaler = StandardScaler()
```

```python
lr = LinearRegression()
pipe = make_pipeline(column_tran, lr)
pipe.fit(x_train, y_train)
pred_lr = pipe.predict(x_test)
r2_score(y_test, pred_lr)
```

```python
lasso = Lasso()
pipe = make_pipeline(column_tran, lasso)
pipe.fit(x_train, y_train)
pred_lasso = pipe.predict(x_test)
r2_score(y_test, pred_lasso)
```

```python
ridge = Ridge()
pipe = make_pipeline(column_tran, ridge)
pipe.fit(x_train, y_train)
pred_ridge = pipe.predict(x_test)
r2_score(y_test, pred_ridge)
```
