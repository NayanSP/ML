```python
def convert(obj):
    lst = []
    for i in ast.literal_eval(obj):
        lst.append(i['name'])
    return lst
```

```python
def convert(obj):
    lst = []
    cnt = 0
    for i in ast.literal_eval(obj):
        if cnt==3:
            break
        lst.append(i['name'])
        cnt+=1
    return lst
```

```python
df['cast'] = df['cast'].apply(convert)
df.head()
```

```python
def fetch_director(obj):
    lst = []
    for i in ast.literal_eval(obj):
        if i['job'] == 'Director':
            lst.append(i['name'])
    return lst
```

```python
df['crew'] = df['crew'].apply(convert)
df.head()
```

```python
df['overview']=df['overview'].apply(lambda x:x.split())
df.head()
```

```python
df['genres'] = df['genres'].apply(lambda x:[i.replace(" ","") for i in x])
df['keywrods'] = df['keywords'].apply(lambda x:[i.replace(" ","") for i in x])
df['cast'] = df['cast'].apply(lambda x:[i.replace(" ","") for i in x])
df['crew'] = df['crew'].apply(lambda x:[i.replace(" ","") for i in x])
df
```

```python
df['tags'] = df.overview + df.genres + df.keywords + df.cast + df.crew
df.head()
```

```python
df1 = df[['movie_id','title','tags']]
df1.tags = df.tags.apply(lambda x:" ".join(x))
df1
```

```python
df1.tags = df1.tags.apply(lambda x:x.lower())
df1
```

```python
df1.tags[0]
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=5000, stop_words='english')
```

```python
vectors = cv.fit_transform(df1['tags']).toarray()
vectors
```

```python
print(cv.get_feature_names_out())
```

```python
import nltk
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def stem(text):
    y = []
    for i in text.split():
        y.append(ps.stem(i))
    
    return " ".join(y)
```

```python
stem(df1.tags[0])
```

```python
df1['tags'] = df1['tags'].apply(stem)
df1.head()
```

```python
vectors = cv.fit_transform(df1['tags']).toarray()
vectors
```

```python
print(cv.get_feature_names_out())
```

```python
vectors[0]
```

```python
from sklearn.metrics.pairwise import cosine_similarity
similar = cosine_similarity(vectors)
```

```python
a = df1[df1['title'] == 'Avatar'].index[0]

distances = similar[a]   # similarity scores for Avatar

movies_list = sorted(
    list(enumerate(distances)),
    reverse=True,
    key=lambda x: x[1]
)[1:6]
movies_list
```

```python
def recommend(moive):
    moive_id = df1[df1['title'] == moive].index[0]
    dist = similar[moive_id]
    movies_list = sorted(
    list(enumerate(dist)),
    reverse=True,
    key=lambda x: x[1]
)[1:6]

    for i in movies_list:
        print(df1.iloc[i[0]].title)
        
```

```python
recommend('Avatar')
```

```python
recommend('Batman Begins')
```
